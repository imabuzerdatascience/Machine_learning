```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
```

```python
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=5,
    n_redundant=2,
    n_classes=2,
    random_state=42
)
```

```python
X.shape
```

### Train_test_split

```python
x_train , x_test , y_train , y_test = train_test_split(X , y , test_size = 0.25 )
```

### Train model using logistic regression 

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report 
model = LogisticRegression()
model.fit(x_train , y_train ) 

y_predict = model.predict(x_test)

report = classification_report(y_test , y_predict)
print("The classification report : " , report )
```

### Train model using gradient boosting

### Train model using gradient boosting 

```python
from sklearn.ensemble import GradientBoostingClassifier 
model = GradientBoostingClassifier(n_estimators = 100)
model.fit(x_train , y_train ) 

y_predict_gb = model.predict(x_test)

report = classification_report(y_test , y_predict_gb)
print("The classification report : " , report )
```

### Train model using XGB 

```python
from xgboost import XGBClassifier 
model = XGBClassifier()
model.fit(x_train , y_train ) 

y_predict_xgb = model.predict(x_test)

report = classification_report(y_test , y_predict_xgb)
print("The classification report : " , report )
```
