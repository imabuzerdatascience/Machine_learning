### import libraries 

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_excel("Raisin_Dataset.xlsx")
df.head()
```

```python
df.Class.unique()
```

There are two class ['Kecimen', 'Besni']

```python
df.shape
```

There are total 900 records and using all the features that we have available, we will build a classification model by using support vector machine 

### Train_test_split

```python
from sklearn.model_selection import train_test_split 
X = df.drop(["Class"] , axis=1)
y = df["Class"]

x_train , x_test , y_train , y_test   = train_test_split(X , y , test_size=0.3 , random_state=12)
x_train.shape
```

### Model Training Using SVM: RBF Kernel

```python
from sklearn.svm import SVC

model = SVC(kernel="rbf")

model.fit(x_train , y_train)

y_predict = model.predict(x_test)

from sklearn.metrics import classification_report
report = classification_report(y_test , y_predict)
print(report)

model.n_iter_
```

### Model Training Using SVM: Linear Kernel

```python
from sklearn.svm import SVC

model = SVC(kernel="linear")

model.fit(x_train , y_train)

y_predict = model.predict(x_test)

from sklearn.metrics import classification_report
report = classification_report(y_test , y_predict)
print(report)

model.n_iter_
```

We used two type of kernel but you can simply see the difference of itreations 

### Scale data 

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train) 

X_train_scaled = scaler.transform(x_train)
X_test_scaled = scaler.transform(x_test)
```

### Model Training Using SVM: RBF Kernel with scaling 

```python
from sklearn.svm import SVC

model = SVC(kernel="rbf")

model.fit(X_train_scaled , y_train)

y_predict = model.predict(X_test_scaled)

from sklearn.metrics import classification_report
report = classification_report(y_test , y_predict)
print(report)

model.n_iter_
```

As you can see above, after scaling the model performance (accuracy, precision, recall) improves. With scaled data training requires a few more iterations but it is not too big, on the other hand precision, recall, accuracy improvements can be very valuable

### Model Training Using SVM: Linear Kernel with scaling 

```python
from sklearn.svm import SVC

model = SVC(kernel="linear")

model.fit(X_train_scaled , y_train)

y_predict = model.predict(X_test_scaled)

from sklearn.metrics import classification_report
report = classification_report(y_test , y_predict)
print(report)

model.n_iter_
```

For linear kernal, scaling the data reduced the number of iterations drastically (down to 708 from 85005907). The model performance on the other hand is more or less the same. There is some difference but nothing major.

### sk learn pipeline

Its help to reduct the size of code . 

```python
from sklearn.pipeline import Pipeline
pipelines = Pipeline([
        ('scaler', StandardScaler()),
        ('svm', SVC(kernel ="rbf"))
])
pipelines.fit(x_test , y_test )
y_pred = pipelines.predict(x_test)

report = classification_report(y_test , y_pred)
print(report)
```

The sklearn Pipeline combines preprocessing and the model into a single workflow, which keeps the code clean, prevents data leakage, and helps build a production‑ready ML pipeline
